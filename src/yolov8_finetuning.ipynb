```python
RUN_DIR = "/content/drive/MyDrive/ua_detrac_runs_ignored_box"
RUN_NAME = "yolov8n"
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data="/content/vehicle-counting/data/yolo/data.yaml",
    epochs=20,
    resume=True,
    imgsz=640,
    batch=64,
    device=0,
    project=RUN_DIR,
    name=RUN_NAME,
    exist_ok=True,
    save_period=1,
    freeze=10
)
```

```python
!./src/preprocessing.py --split test
!./src/convert_to_yolo.py --split test
```

```python
from ultralytics import YOLO

model = YOLO(
    f"{RUN_DIR}/{RUN_NAME}/weights/best.pt"
)

metrics = model.val(
    project="/content/drive/MyDrive/yolo_results",
    name="yolov8n_detrac_test",
    data="/content/vehicle-counting/data/yolo/data.yaml",
    split="test",
    imgsz=640,
    batch=64,
    device=0,
    conf=0.494
)

print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall: {metrics.box.mr:.4f}")
```
